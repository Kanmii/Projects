```python
import re

# Function to slice the email into username and domain
def email_slicer(email):
    # Convert email to lowercase for consistency
    email = email.lower()

    # Validate using regex for proper format: username@domain.tld
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    if re.match(pattern, email):
        # Split the email into username and domain
        username, domain = email.split('@')
        return username, domain
    else:
        # Return None if the email format is invalid
        return None, None

# Function to categorize domain type
def categorize_domain(domain):
    # List of common personal email providers
    personal_domains = ['gmail.com', 'yahoo.com', 'outlook.com', 'hotmail.com', 'aol.com']
    if domain in personal_domains:
        return 'Personal Email'
    else:
        return 'Corporate/Other Email'

# Get email input from the user
email = input("Enter your email address: ")

# Call the function to slice the email
username, domain = email_slicer(email)

# Display the result
if username and domain:
    # Domain categorization
    domain_type = categorize_domain(domain)

    # Print output with good formatting
    print("\n📧 Email Details:")
    print(f"Username       : {username}")
    print(f"Domain         : {domain}")
    print(f"Domain Type    : {domain_type}")
else:
    print("❌ Invalid email format! Please enter a valid email like 'name@example.com'.")
```
