```python
# Simple & Compound Interest Calculator

# Function to calculate Simple Interest
def calculate_simple_interest(P, R, T):
    # SI = (Principal * Rate * Time)
    return (P * R * T)

# Function to calculate Compound Interest
def calculate_compound_interest(P, R, T, n):
    # Compound Interest = P * (1 + R/n)^(n*T) - P
    amount = P * (1 + R / n) ** (n * T)
    return amount - P

# Take validated user input
try:
    P = float(input("Enter principal amount (₦): "))
    if P < 0:
        raise ValueError("Principal cannot be negative.")

    R = float(input("Enter annual interest rate (as decimal, e.g., 0.05 for 5%): "))
    if R < 0:
        raise ValueError("Rate cannot be negative.")

    T = float(input("Enter time period in years: "))
    if T < 0:
        raise ValueError("Time period cannot be negative.")

    print("\nChoose interest type:")
    print("1. Simple Interest")
    print("2. Compound Interest")
    interest_type = input("Enter 1 or 2: ")

    if interest_type == "1":
        # Simple Interest calculation
        SI = calculate_simple_interest(P, R, T)
        print(f"\n📌 Simple Interest: ₦{SI:.2f}")

    elif interest_type == "2":
        print("\nChoose compounding frequency:")
        print("1. Yearly")
        print("2. Quarterly")
        print("3. Monthly")
        freq_choice = input("Enter 1, 2, or 3: ")

        # Convert choice to compounding frequency (n)
        if freq_choice == "1":
            n = 1  # Yearly
        elif freq_choice == "2":
            n = 4  # Quarterly
        elif freq_choice == "3":
            n = 12  # Monthly
        else:
            raise ValueError("Invalid compounding frequency.")

        CI = calculate_compound_interest(P, R, T, n)
        print(f"\n📌 Compound Interest: ₦{CI:.2f}")

    else:
        print("Invalid interest type selected.")

except ValueError as ve:
    print(f"❌ Input Error: {ve}")
```
