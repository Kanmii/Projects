```python
import nltk
import requests
from nltk.tokenize import word_tokenize

# Download NLTK tokenizer data (first-time only)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True) # Download the missing 'punkt_tab' resource

# Dictionary of predefined responses (expanded)
responses = {
    "hi": "Hello! How can I assist you?",
    "hello": "Hi there! How can I help you?",
    "how are you": "I'm a bot, but I'm here to help!",
    "what is python": "Python is a powerful programming language used in web development, data science, and more.",
    "where are you learning data science": "At Sail Innovation Lab!",
    "what programming language are you learning?": "Python!",
    "bye": "Goodbye! Have a great day!",
    "what's your name": "I'm your friendly chatbot assistant.",
    "thank you": "You're welcome!",
    "thanks": "Anytime!"
}

# Function to fetch real-time weather using an API
def get_weather(city):
    api_key = "f46a48b1c1779ffae52b2f57231e8f45"  # Replace with your actual API key
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"
    try:
        response = requests.get(url)
        data = response.json()
        if data.get("main"):
            temp = data['main']['temp']
            desc = data['weather'][0]['description']
            return f"The weather in {city.title()} is {desc} with a temperature of {temp}°C."
        else:
            return "Sorry, I couldn't fetch the weather for that location."
    except:
        return "Weather service is temporarily unavailable."

# Function to handle user input using NLP
def generate_response(user_input):
    tokens = word_tokenize(user_input.lower())

    # Check for weather request
    if "weather" in tokens:
        for i, token in enumerate(tokens):
            if token == "in" and i + 1 < len(tokens):
                city = tokens[i + 1]
                return get_weather(city)
        return "Please specify a city, like: 'weather in Lagos'."

    # Try matching exact query from predefined dictionary
    if user_input in responses:
        return responses[user_input]

    # Fallback default
    return "I'm sorry, I don't understand that yet."

# Infinite loop to keep chatbot running
while True:
    try:
        user_input = input("You: ").strip().lower()

        # Get response
        bot_response = generate_response(user_input)

        # Print bot response
        print(f"Chatbot: {bot_response}")

        # Exit on bye
        if user_input == "bye":
            break

    except KeyboardInterrupt:
        print("Chatbot interrupted!, Goodbye")
        break
```
