```python
# Simple Unit Converter with many options and easy code structure

# Define a helper function to perform regular conversions
def convert(value, factor, operation="*"):
    # Multiply or divide based on the operation
    return round(value * factor, 3) if operation == "*" else round(value / factor, 3)

# Dictionary to manage all conversions: name, factor, and direction
conversion_map = {
    "1": {"from": "meters", "to": "kilometers", "factor": 1000, "operation": "/"},
    "2": {"from": "kilometers", "to": "meters", "factor": 1000, "operation": "*"},
    "3": {"from": "kilometers", "to": "miles", "factor": 0.621371, "operation": "*"},
    "4": {"from": "miles", "to": "kilometers", "factor": 0.621371, "operation": "/"},
    "5": {"from": "celsius", "to": "fahrenheit", "custom": lambda c: round((c * 9 / 5) + 32, 3)},
    "6": {"from": "fahrenheit", "to": "celsius", "custom": lambda f: round((f - 32) * 5 / 9, 3)},
    "7": {"from": "grams", "to": "kilograms", "factor": 1000, "operation": "/"},
    "8": {"from": "kilograms", "to": "grams", "factor": 1000, "operation": "*"},
    "9": {"from": "km/hr", "to": "miles/hr", "factor": 0.621371, "operation": "*"},
    "10": {"from": "miles/hr", "to": "km/hr", "factor": 0.621371, "operation": "/"},
    "11": {"from": "liters", "to": "milliliters", "factor": 1000, "operation": "*"},
    "12": {"from": "milliliters", "to": "liters", "factor": 1000, "operation": "/"},
}

# Display menu to the user
print("\n🔄 Unit Converter Menu")
for key, val in conversion_map.items():
    print(f"{key}. {val['from'].capitalize()} to {val['to'].capitalize()}")

# Ask for user's choice
choice = input("Enter your choice (1-12): ")

# Check if choice is valid
if choice in conversion_map:
    # Ask for the value to convert
    value = float(input(f"Enter value in {conversion_map[choice]['from']}: "))

    # Fetch the selected conversion details
    conv = conversion_map[choice]

    # If the conversion is a custom one (like temperature)
    if "custom" in conv:
        result = conv["custom"](value)
    else:
        result = convert(value, conv["factor"], conv["operation"])

    # Show result
    print(f"✅ {value} {conv['from']} = {result} {conv['to']}")
else:
    print("❌ Invalid input! Please enter a number between 1 and 12.")
```
