```python
import datetime
import json
import os

# File to store birthdays
FILE_PATH = "birthdays.json"

# Load existing birthdays or return default ones
def load_birthdays():
    if os.path.exists(FILE_PATH):
        with open(FILE_PATH, "r") as file:
            return json.load(file)
    else:
        return {
            "Sunday": "2012-03-15",
            "Victor": "1999-07-10",
            "Bob": "2004-12-25",
            "Aliyah": "2001-05-02",
            "John": "2022-03-15",
            "Esther": "2014-05-12"
        }

# Save updated birthday dictionary to file
def save_birthdays(birthdays):
    with open(FILE_PATH, "w") as file:
        json.dump(birthdays, file)

# Adds suffix to day numbers like "st", "nd", "rd"
def day_suffix(day):
    if 11 <= day <= 13:
        return 'th'
    last = day % 10
    return {1: 'st', 2: 'nd', 3: 'rd'}.get(last, 'th')

# Load birthdays
birthdays = load_birthdays()

# Get today's date
today = datetime.date.today()
today_month = today.month
today_day = today.day

# Check and display reminders
print(f"Today is {today.strftime('%A, %B %d, %Y')}")
print("📌 Birthday Reminders:")
for name, date_str in birthdays.items():
    bday = datetime.datetime.strptime(date_str, "%Y-%m-%d").date()
    if bday.month == today_month and bday.day == today_day:
        print(f"🎉 Reminder: {name}'s birthday is today!")
    elif bday.month == today_month:
        suffix = day_suffix(bday.day)
        formatted = bday.strftime(f"%B {bday.day}{suffix}")
        print(f"🗓️ Upcoming: {name}'s birthday is on {formatted}")

# Option to add new birthdays
add_new = input("\nDo you want to add a new birthday? (yes/no): ").strip().lower()
if add_new == "yes":
    name = input("Enter the name: ").strip()
    bday_input = input("Enter birthday (YYYY-MM-DD): ").strip()
    try:
        datetime.datetime.strptime(bday_input, "%Y-%m-%d")  # Validate format
        birthdays[name] = bday_input
        save_birthdays(birthdays)
        print(f"✅ {name}'s birthday saved!")
    except ValueError:
        print("❌ Invalid date format. Use YYYY-MM-DD.")
```
