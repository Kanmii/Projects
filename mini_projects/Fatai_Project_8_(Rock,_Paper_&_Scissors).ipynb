```python
import random

# Define basic and advanced choices
basic_choices = ["rock", "paper", "scissors"]
advanced_choices = ["rock", "paper", "scissors", "lizard", "spock"]

# Define win conditions using dictionary for easy lookup
win_rules = {
    "rock": ["scissors", "lizard"],
    "paper": ["rock", "spock"],
    "scissors": ["paper", "lizard"],
    "lizard": ["spock", "paper"],
    "spock": ["scissors", "rock"]
}

# Function to determine winner
def determine_winner(user, computer):
    if user == computer:
        return "draw"
    elif computer in win_rules[user]:
        return "user"
    else:
        return "computer"

# Start game with options
print("🎮 Welcome to Rock, Paper, Scissors Game!")
mode = input("Choose mode: (1) Basic (2) Advanced: ").strip()

# Determine choices based on mode
if mode == "2":
    choices = advanced_choices
    print("🧠 Advanced mode activated: rock, paper, scissors, lizard, spock")
else:
    choices = basic_choices
    print("🧩 Basic mode: rock, paper, scissors")

# Initialize scores
user_score = 0
computer_score = 0
round_num = 1

# Start game loop
while True:
    print(f"\n🔁 Round {round_num}")
    user_choice = input("Enter your choice or type 'exit' to quit: ").lower().strip()

    # Allow user to quit
    if user_choice == "exit":
        print("\n🚪 Exiting game...")
        break

    # Validate input
    if user_choice not in choices:
        print("❌ Invalid input! Please choose from:", ", ".join(choices))
        continue

    # Generate computer choice
    computer_choice = random.choice(choices)
    print(f"🤖 Computer chose: {computer_choice}")

    # Determine winner
    result = determine_winner(user_choice, computer_choice)

    # Update scores
    if result == "user":
        user_score += 1
        print("✅ You win this round!")
    elif result == "computer":
        computer_score += 1
        print("❌ Computer wins this round!")
    else:
        print("🤝 It's a draw!")

    # Display scores
    print(f"📊 Score => You: {user_score} | Computer: {computer_score}")
    round_num += 1

# Final message
print("\n🏁 Final Score:")
print(f"🙋‍♂️ You: {user_score} | 🤖 Computer: {computer_score}")
print("🎉 Thanks for playing!")
```
