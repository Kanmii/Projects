```python
import time
from tqdm import tqdm
import sys
import platform

# Optional sound function for different OS
def beep():
    if platform.system() == "Windows":
        import winsound
        winsound.Beep(1000, 500)
    else:
        print('\a')  # Bell character for Unix/macOS (may not work on all terminals)

# Get countdown time from user in seconds
seconds = int(input("Enter countdown time in seconds: "))

# Main countdown logic
try:
    for remaining in tqdm(range(seconds, 0, -1), desc="Counting down", unit="s"):
        # Convert remaining seconds to minutes:seconds format
        mins, secs = divmod(remaining, 60)
        time_display = f"Time left: {mins:02d}:{secs:02d}"
        print(time_display, end="\r")  # Overwrite same line
        time.sleep(1)
except KeyboardInterrupt:
    print("\n⏸️ Countdown paused. Press Enter to resume...")
    input()
    # Resume countdown
    for remaining in tqdm(range(remaining, 0, -1), desc="Resuming", unit="s"):
        mins, secs = divmod(remaining, 60)
        print(f"Time left: {mins:02d}:{secs:02d}", end="\r")
        time.sleep(1)

# When countdown completes
print("\nTime's up!!!! 🎉")
beep()
```
