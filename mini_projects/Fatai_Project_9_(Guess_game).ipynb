```python
import random

# Function to return max attempts and step based on difficulty
def set_difficulty(level):
    if level == 'easy':
        return 10, 10  # Easy: More attempts, wider range steps
    elif level == 'medium':
        return 7, 5    # Medium: Moderate
    else:
        return 5, 1    # Hard: Few attempts, closer guesses

# Function to play one round
def play_round(level, best_score):
    max_attempts, step = set_difficulty(level)
    number = random.randrange(step, 101, step)  # Random target number
    attempts = 0  # Track number of guesses

    print(f"\n🎯 Guess the number (multiples of {step}) between {step} and 100")
    print(f"You have {max_attempts} attempts")

    while attempts < max_attempts:
        try:
            guess = int(input("Enter your guess: "))
        except ValueError:
            print("❌ Invalid input! Please enter a number.")
            continue

        attempts += 1

        if guess < number:
            print("🔻 Too low!")
        elif guess > number:
            print("🔺 Too high!")
        else:
            print(f"✅ Correct! You guessed in {attempts} attempt(s).")

            # Update best score
            if best_score is None or attempts < best_score:
                best_score = attempts
                print("🏆 New best score!")
            return best_score

    print(f"💀 You've used all {max_attempts} attempts. The number was {number}.")
    return best_score

# Main game loop
def main():
    best_score = None
    print("🎮 Welcome to the Number Guessing Game!")

    while True:
        # Difficulty selection
        level = input("\nChoose difficulty (easy / medium / hard): ").lower().strip()
        if level not in ["easy", "medium", "hard"]:
            print("❌ Invalid difficulty. Try again.")
            continue

        # Play one round
        best_score = play_round(level, best_score)

        # Show current best score
        if best_score:
            print(f"📊 Best score so far: {best_score} attempt(s)")

        # Ask to play again
        replay = input("\n🔁 Play again? (yes / no): ").lower().strip()
        if replay != "yes":
            print("👋 Thanks for playing!")
            break

# Start the game
main()
```
