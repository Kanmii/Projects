```python
import random

# Welcome message
print("🎉 Welcome to the Mad Libs Game! 🎉")

# Give the user theme options
print("\nChoose a story theme:")
print("1. Adventure")
print("2. Mystery")
print("3. Fantasy")

# Ask the user to pick a theme
theme_choice = input("Enter the number of your choice (1-3): ")

# Ask user for words to insert into the story
# These will be used based on the chosen theme
noun = input("Enter a noun (thing): ")
verb = input("Enter a verb (action): ")
adjective = input("Enter an adjective (describing word): ")
place = input("Enter a place: ")
animal = input("Enter an animal: ")
exclamation = input("Enter an exclamation (e.g., Wow, Oh no): ")

# Define 3 different story templates
stories = {
    "1": f"On a sunny day in {place}, a {adjective} {noun} decided to {verb}. Suddenly, a wild {animal} appeared! \"{exclamation}!\" shouted the {noun}. And thus began the greatest adventure of all time.",
    "2": f"It was a dark night in {place}. A {adjective} {noun} was investigating a strange noise. As it tried to {verb}, it stumbled upon a clue left by a {animal}. \"{exclamation}!\" it whispered.",
    "3": f"In the magical land of {place}, a {adjective} {noun} had one goal: to {verb} the mighty {animal}. With courage in its heart, it screamed, \"{exclamation}!\" and charged into battle."
}

# Select story based on user input
selected_story = stories.get(theme_choice, "Invalid choice. Please restart the game and pick 1, 2, or 3.")

# Print the result
print("\n📝 Here is your Mad Libs story:")
print(selected_story)

# Save story to a file
with open("mad_libs_story.txt", "w") as file:
    file.write(selected_story)

print("\n✅ Your story has been saved as 'mad_libs_story.txt'.")
```
